```python
import mysql.connector
import datetime
def conn():
    return mysql.connector.connect(
        host='localhost',
        user='root',
        password='Sujay@2002',
        database='expense_tracker')
def add_expense():
    category=input("Enter category like medicines,grocery....")
    amount=float(input("Enter amount spent"))
    date_str=input("Enter date in format yyyy-mm-dd if not today")
    description=input("Enter description (optional)")
    #getting system date or converting entered date in required format
    date=date_str if date_str else datetime.datetime.today().strftime('%Y-%m-%d') 
    #strftime-function of datetime which converts date in given format
    sql='''insert into expense(expense_dt,category,amount,descript)
    values (%s,%s,%s,%s)'''
    vals=(date,category,amount,description)

    con=conn()
    cursor=con.cursor()
    cursor.execute(sql,vals)
    con.commit()
    print("Added successfully")
    con.close()
def delete_expense():
    sr_no=int(input("Enter ID of expense to be deleted"))
    con=conn()
    cursor=con.cursor()
    cursor.execute("delete from expense where sr_no=%s",(sr_no,))
    con.commit()
    print("deleted Sucessfully")
    con.close()
def Summary():
    month=int(input("Enter month as a number between 1 and 12"))
    year=int(input("Enter year in format yyyy"))
    if(month<1 or month>12):
        print("Invalid month")
    else:
        con=conn()
        cursor=con.cursor()
        sql='''select sum(amount),category from expense 
        where month(expense_dt)=%s and year(expense_dt)=%s
        group by category'''
        cursor.execute(sql,(month,year))
        records=cursor.fetchall()
        print("Total spent | category")
        for row in records:
            print(f"{row[0]} | {row[1]}")
        con.close()
  
def main():
    while True:
        print("-------Track Your Expenses--------")
        print("1.Add a new expense")
        print("2.Delete an expense on ID")
        print("3.View all expenses")
        print("4.View monthly summary")
        print("5.Exit\n")
        ch=int(input("Enter your choice (1-5)"))
        if ch==1:
            add_expense()
        elif ch==2:
            delete_expense()
        elif ch==4:
            Summary()
            
        elif ch==5:
            print("Thank you for using tracker!!")
            break
        else:
            print("Wrong choice!! Please retry")
if __name__=="__main__":
    main()
```
